```python
#Python Data types 
#basic data types : int, float, string( str), bool, none

#integer (int); whole value without decimal point. 
age=20
print( age, type (age))
# Float (float ): number with decimal point 
x= 3.5
print(x, type(x))
#string (str): textual information 
name= "astha maurya"
print(name, type(name))
#boolean (bool ): logical values True or False
is_active= True
is_equalto= False
print(is_active, is_equalto, type(is_equalto))
#None (nonetype):Represents the absence of a value.
project_status=None
print(project_status, type(project_status))
```

```python
#Dynamic Typeing : Pyhton is dynamically typed, it means you don't need to explicitly define the variable date type. Python defines it at a runtime.

x = 100

print(type(x))

x = "Hello"

print(type(x))

#The same variable name now refers to a different type of object.
```

```python
""" 
Static typing vs dynamic typing

Static typing	

Types are checked before execution or during compilation	
Java, C++	
More explicit type declarations are common	


Dynamic typing

Types are associated with objects and checked at runtime
Python, JavaScript
Explicit declarations are generally optional

"""
```

```python
#Type conversion means converting one data type into another.

age = "30"

print(type(age))

age = int(age)

print(type(age))
```

```python
print(int("100"))       # 100
print(float("10.5"))    # 10.5
print(str(500))         # "500"
print(bool(1))          # True
print(bool(0))          # False
print(bool(""))         # False
print(bool("False"))    # True

# Why does bool("False") return True?

# Because Python considers every non-empty string to be truthy, regardless of its textual meaning
```
